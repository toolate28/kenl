## 🔴 Red Team Functions

Offensive security operations for vulnerability assessment and penetration testing.

### Network Reconnaissance

```python
def network_reconnaissance(target_network="192.168.1.0/24"):
    """
    Perform network reconnaissance using nmap-style analysis
    
    Args:
        target_network: CIDR notation network range
    
    Returns:
        DataFrame with discovered hosts and services
    """
    atom_tag = generate_atom_tag("RECON")
    print(f"🔴 Red Team: Network Reconnaissance")
    print(f"🏷️  ATOM Tag: {atom_tag}")
    print(f"🎯 Target: {target_network}")
    
    # Simulated reconnaissance data (replace with actual nmap integration)
    hosts = [
        {"ip": "192.168.1.1", "hostname": "gateway", "os": "Linux", "risk": "Medium"},
        {"ip": "192.168.1.10", "hostname": "workstation-01", "os": "Windows", "risk": "High"},
        {"ip": "192.168.1.20", "hostname": "server-01", "os": "Linux", "risk": "Critical"},
    ]
    
    df = pd.DataFrame(hosts)
    
    # Visualize risk distribution
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))
    
    # Risk level counts
    risk_counts = df['risk'].value_counts()
    colors = {'Low': '#51cf66', 'Medium': '#ffd43b', 'High': '#ff6b6b', 'Critical': '#c92a2a'}
    ax1.bar(risk_counts.index, risk_counts.values, color=[colors.get(r, '#999') for r in risk_counts.index])
    ax1.set_title('🎯 Risk Distribution')
    ax1.set_xlabel('Risk Level')
    ax1.set_ylabel('Count')
    
    # OS distribution
    os_counts = df['os'].value_counts()
    ax2.pie(os_counts.values, labels=os_counts.index, autopct='%1.1f%%', startangle=90)
    ax2.set_title('🖥️ OS Distribution')
    
    plt.tight_layout()
    plt.show()
    
    return df

# Example usage
# recon_results = network_reconnaissance("192.168.1.0/24")
# display(recon_results)

print("✅ Red Team reconnaissance function ready")
```

### Vulnerability Assessment

```python
def vulnerability_assessment(host_data):
    """
    Assess vulnerabilities for discovered hosts
    
    Args:
        host_data: DataFrame with host information
    
    Returns:
        DataFrame with vulnerability assessment results
    """
    atom_tag = generate_atom_tag("VULN")
    print(f"🔴 Red Team: Vulnerability Assessment")
    print(f"🏷️  ATOM Tag: {atom_tag}")
    
    # Simulated vulnerability data
    vulnerabilities = [
        {"cve": "CVE-2024-0001", "severity": "Critical", "cvss": 9.8, "host": "192.168.1.20", "service": "Apache 2.4.41"},
        {"cve": "CVE-2024-0002", "severity": "High", "cvss": 7.5, "host": "192.168.1.10", "service": "SMBv1"},
        {"cve": "CVE-2024-0003", "severity": "Medium", "cvss": 5.3, "host": "192.168.1.1", "service": "OpenSSH 7.9"},
    ]
    
    df = pd.DataFrame(vulnerabilities)
    
    # Vulnerability heat map
    fig, ax = plt.subplots(figsize=(12, 4))
    pivot = df.pivot_table(values='cvss', index='host', columns='severity', aggfunc='mean', fill_value=0)
    sns.heatmap(pivot, annot=True, fmt='.1f', cmap='RdYlGn_r', ax=ax, cbar_kws={'label': 'CVSS Score'})
    ax.set_title('🔥 Vulnerability Heat Map by Host')
    plt.tight_layout()
    plt.show()
    
    return df

# Example usage
# vuln_results = vulnerability_assessment(recon_results)
# display(vuln_results)

print("✅ Vulnerability assessment function ready")
```

## 🔵 Blue Team Functions

Defensive security operations for monitoring, detection, and incident response.

### Log Analysis

```python
def analyze_security_logs(log_file=None, timeframe_hours=24):
    """
    Analyze security logs for suspicious activity
    
    Args:
        log_file: Path to log file (optional)
        timeframe_hours: Analysis window in hours
    
    Returns:
        DataFrame with security events and analysis
    """
    atom_tag = generate_atom_tag("LOGANALYSIS")
    print(f"🔵 Blue Team: Log Analysis")
    print(f"🏷️  ATOM Tag: {atom_tag}")
    print(f"⏱️  Timeframe: Last {timeframe_hours} hours")
    
    # Simulated log events
    now = datetime.now()
    events = []
    for i in range(100):
        event_time = now - timedelta(hours=np.random.randint(0, timeframe_hours))
        events.append({
            "timestamp": event_time,
            "event_type": np.random.choice(["login_success", "login_failure", "port_scan", "file_access", "privilege_escalation"]),
            "source_ip": f"192.168.1.{np.random.randint(1, 255)}",
            "severity": np.random.choice(["Low", "Medium", "High", "Critical"], p=[0.6, 0.25, 0.1, 0.05])
        })
    
    df = pd.DataFrame(events)
    df = df.sort_values('timestamp')
    
    # Timeline visualization
    fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(14, 8))
    
    # Events over time
    df.set_index('timestamp').resample('1H').size().plot(kind='line', ax=ax1, color='#4dabf7', marker='o')
    ax1.set_title('📈 Security Events Timeline')
    ax1.set_xlabel('Time')
    ax1.set_ylabel('Event Count')
    ax1.grid(True, alpha=0.3)
    
    # Event type distribution
    event_counts = df['event_type'].value_counts()
    ax2.barh(event_counts.index, event_counts.values, color='#845ef7')
    ax2.set_title('📊 Event Type Distribution')
    ax2.set_xlabel('Count')
    
    plt.tight_layout()
    plt.show()
    
    # Identify suspicious patterns
    suspicious = df[df['event_type'].isin(['port_scan', 'privilege_escalation'])]
    print(f"\n⚠️  Suspicious events detected: {len(suspicious)}")
    
    return df

# Example usage
# log_analysis = analyze_security_logs(timeframe_hours=24)

print("✅ Log analysis function ready")
```

### Incident Response Dashboard

```python
def incident_response_dashboard(incident_data):
    """
    Generate incident response dashboard with key metrics
    
    Args:
        incident_data: DataFrame with incident information
    
    Returns:
        Visualization dashboard
    """
    atom_tag = generate_atom_tag("INCIDENT")
    print(f"🔵 Blue Team: Incident Response Dashboard")
    print(f"🏷️  ATOM Tag: {atom_tag}")
    
    # Create summary metrics
    total_incidents = len(incident_data)
    critical_incidents = len(incident_data[incident_data['severity'] == 'Critical'])
    high_incidents = len(incident_data[incident_data['severity'] == 'High'])
    
    # Display key metrics
    display(HTML(f"""
    <div style="background: #1e1e1e; padding: 20px; border-radius: 10px; color: white;">
        <h2>🚨 Incident Response Metrics</h2>
        <div style="display: flex; justify-content: space-around; margin-top: 20px;">
            <div style="text-align: center;">
                <h1 style="color: #4dabf7;">{total_incidents}</h1>
                <p>Total Incidents</p>
            </div>
            <div style="text-align: center;">
                <h1 style="color: #c92a2a;">{critical_incidents}</h1>
                <p>Critical</p>
            </div>
            <div style="text-align: center;">
                <h1 style="color: #ff6b6b;">{high_incidents}</h1>
                <p>High</p>
            </div>
        </div>
    </div>
    """))
    
    return None

# Example usage
# incident_response_dashboard(log_analysis)

print("✅ Incident response dashboard ready")
```

## 🟣 Purple Team Functions

Integrated security validation combining offensive and defensive perspectives.

### Attack Simulation & Validation

```python
def purple_team_assessment(red_team_data, blue_team_data):
    """
    Purple team assessment combining red and blue team data
    
    Args:
        red_team_data: Vulnerability assessment results
        blue_team_data: Log analysis results
    
    Returns:
        Comprehensive security posture assessment
    """
    atom_tag = generate_atom_tag("PURPLE")
    print(f"🟣 Purple Team: Security Posture Assessment")
    print(f"🏷️  ATOM Tag: {atom_tag}")
    
    # Security posture metrics
    metrics = {
        "attack_surface": 85,  # 0-100 scale
        "detection_coverage": 72,
        "response_time": 15,  # minutes
        "vulnerability_count": len(red_team_data) if red_team_data is not None else 0,
        "incident_count": len(blue_team_data) if blue_team_data is not None else 0
    }
    
    # Radar chart for security posture
    categories = ['Attack Surface\nReduction', 'Detection\nCoverage', 'Response\nSpeed', 
                  'Vulnerability\nManagement', 'Threat\nIntelligence']
    values = [65, 72, 80, 58, 70]
    
    angles = np.linspace(0, 2 * np.pi, len(categories), endpoint=False).tolist()
    values += values[:1]
    angles += angles[:1]
    
    fig, ax = plt.subplots(figsize=(8, 8), subplot_kw=dict(projection='polar'))
    ax.plot(angles, values, 'o-', linewidth=2, color='#845ef7')
    ax.fill(angles, values, alpha=0.25, color='#845ef7')
    ax.set_xticks(angles[:-1])
    ax.set_xticklabels(categories)
    ax.set_ylim(0, 100)
    ax.set_title('🟣 Security Posture Assessment', pad=20, fontsize=14, fontweight='bold')
    ax.grid(True)
    
    plt.tight_layout()
    plt.show()
    
    # Recommendations
    print("\n📋 Purple Team Recommendations:")
    print("1. ⬆️  Increase detection coverage in DMZ zone")
    print("2. 🔧 Patch critical vulnerabilities within 24 hours")
    print("3. 📚 Enhance security awareness training")
    print("4. 🔄 Implement automated threat hunting")
    
    return metrics

# Example usage
# purple_assessment = purple_team_assessment(vuln_results, log_analysis)

print("✅ Purple team assessment function ready")
```

## 📊 Advanced Visualizations

### Network Topology Mapping

```python
def visualize_network_topology(hosts_data):
    """
    Create network topology visualization using NetworkX
    
    Args:
        hosts_data: DataFrame with host information
    
    Returns:
        Network graph visualization
    """
    atom_tag = generate_atom_tag("TOPOLOGY")
    print(f"📊 Visualization: Network Topology")
    print(f"🏷️  ATOM Tag: {atom_tag}")
    
    # Create network graph
    G = nx.Graph()
    
    # Add nodes (simulated network)
    nodes = [
        ("Gateway", {"type": "gateway", "risk": "medium"}),
        ("Server-01", {"type": "server", "risk": "critical"}),
        ("Server-02", {"type": "server", "risk": "high"}),
        ("Workstation-01", {"type": "workstation", "risk": "low"}),
        ("Workstation-02", {"type": "workstation", "risk": "medium"}),
        ("IoT-01", {"type": "iot", "risk": "high"}),
    ]
    
    G.add_nodes_from(nodes)
    
    # Add edges (connections)
    edges = [
        ("Gateway", "Server-01"),
        ("Gateway", "Server-02"),
        ("Gateway", "Workstation-01"),
        ("Gateway", "Workstation-02"),
        ("Gateway", "IoT-01"),
        ("Server-01", "Server-02"),
    ]
    
    G.add_edges_from(edges)
    
    # Color mapping by risk
    risk_colors = {
        "low": "#51cf66",
        "medium": "#ffd43b",
        "high": "#ff6b6b",
        "critical": "#c92a2a"
    }
    
    node_colors = [risk_colors[G.nodes[node]['risk']] for node in G.nodes()]
    
    # Draw network
    plt.figure(figsize=(12, 10))
    pos = nx.spring_layout(G, k=0.5, iterations=50)
    
    nx.draw_networkx_nodes(G, pos, node_color=node_colors, node_size=2000, alpha=0.9)
    nx.draw_networkx_labels(G, pos, font_size=10, font_weight='bold')
    nx.draw_networkx_edges(G, pos, width=2, alpha=0.5, edge_color='#4dabf7')
    
    plt.title('🌐 Network Topology Map (Risk Assessment)', fontsize=16, fontweight='bold', pad=20)
    plt.axis('off')
    
    # Add legend
    legend_elements = [
        plt.Line2D([0], [0], marker='o', color='w', markerfacecolor='#51cf66', markersize=10, label='Low Risk'),
        plt.Line2D([0], [0], marker='o', color='w', markerfacecolor='#ffd43b', markersize=10, label='Medium Risk'),
        plt.Line2D([0], [0], marker='o', color='w', markerfacecolor='#ff6b6b', markersize=10, label='High Risk'),
        plt.Line2D([0], [0], marker='o', color='w', markerfacecolor='#c92a2a', markersize=10, label='Critical Risk'),
    ]
    plt.legend(handles=legend_elements, loc='upper left', fontsize=10)
    
    plt.tight_layout()
    plt.show()
    
    return G

# Example usage
# network_graph = visualize_network_topology(recon_results)

print("✅ Network topology visualization ready")
```

## 🏷️ ATOM Integration

### ATOM Trail Logger

```python
def log_atom_trail(operation, details, result="success"):
    """
    Log security operation to ATOM trail
    
    Args:
        operation: Type of operation (RECON, VULN, INCIDENT, etc.)
        details: Operation details and intent
        result: Operation result (success/failure)
    
    Returns:
        ATOM tag
    """
    atom_tag = generate_atom_tag(operation)
    timestamp = datetime.now().strftime('%Y-%m-%d %H:%M:%S')
    
    # Create ATOM trail entry
    atom_entry = {
        "atom_tag": atom_tag,
        "timestamp": timestamp,
        "operation": operation,
        "details": details,
        "result": result,
        "module": "KENL8-security"
    }
    
    # Display ATOM entry
    display(HTML(f"""
    <div style="background: #2d2d2d; padding: 15px; border-left: 4px solid #845ef7; margin: 10px 0;">
        <h3 style="color: #845ef7; margin: 0;">🏷️ ATOM Trail Entry</h3>
        <p style="margin: 5px 0;"><strong>Tag:</strong> {atom_tag}</p>
        <p style="margin: 5px 0;"><strong>Operation:</strong> {operation}</p>
        <p style="margin: 5px 0;"><strong>Time:</strong> {timestamp}</p>
        <p style="margin: 5px 0;"><strong>Details:</strong> {details}</p>
        <p style="margin: 5px 0;"><strong>Result:</strong> <span style="color: {'#51cf66' if result == 'success' else '#ff6b6b'};">{result}</span></p>
    </div>
    """))
    
    # In production, write to ATOM log file
    # with open(os.path.expanduser(os.path.join(os.environ.get('KENL_HOME', os.path.expanduser('~/.kenl')), 'atom-trail.log')), 'a') as f:
    #     f.write(json.dumps(atom_entry) + '\n')
    
    return atom_tag

# Example usage
# atom_tag = log_atom_trail(
#     operation="RECON",
#     details="Network reconnaissance of 192.168.1.0/24 completed. Identified 3 hosts with security concerns.",
#     result="success"
# )

print("✅ ATOM trail logger ready")
```
